```python
#reverse a string
s='aabbbaabbcccs'
s=list(s)
n=len(s)
start=0
end=n-1
while(start<end):
    s[start],s[end]=s[end],s[start]
    start+=1
    end-=1
s=''.join(s)
print(s)
```

```python
#palindrome

s='madam'
s=list(s)
n=len(s)
flag=True
start=0
end=n-1
while(start<end):
    if s[start]!=s[end]:
        flag=False
        print("not a palindrome")
        break
    
    start+=1
    end-=1

if flag:
    print("palindrome")
  
```

```python
s='1.1.1.1'
i=0
n=len(s)
ans=''
while(i<n):
    if s[i]=='.':
        ans+='[.]'
    else:
        ans+=s[i]
    i+=1

print(ans)
```

```python
#rotate string in anticlockwise direction
s='amazon'
n=len(s)
s=list(s)
temp=s[0]
for i in range(1,n):
    s[i-1]=s[i]

s[n-1]=temp
s=''.join(s)
print(s)

```
