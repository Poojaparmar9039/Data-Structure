```python
#reverse a string
s='aabbbaabbcccs'
s=list(s)
n=len(s)
start=0
end=n-1
while(start<end):
    s[start],s[end]=s[end],s[start]
    start+=1
    end-=1
s=''.join(s)
print(s)
```

```python
#palindrome

s='madam'
s=list(s)
n=len(s)
flag=True
start=0
end=n-1
while(start<end):
    if s[start]!=s[end]:
        flag=False
        print("not a palindrome")
        break
    
    start+=1
    end-=1

if flag:
    print("palindrome")
  
```

```python
#check if string is pangram
s="thequickbrownfoxjumpsoverthelazydog"
n=len(s)
ans=[0]*26
for i in range(n):
    indx=ord(s[i])-ord('a')
    ans[indx]=1

flag=True
for i in range(26):
    if(ans[i]==0):
        flag=False
        print("not a pangram")
        break

if flag:
    print("pangram")
```

```python
# longest substring without repeating characters
s = 'abcdefcbeadbf'
s = list(s)
n = len(s)

count = [0] * 26
length = 0
l = 0
r = 0

while r < n:
    while count[ord(s[r]) - ord('a')]:
        count[ord(s[l]) - ord('a')] = 0
        l += 1

    count[ord(s[r]) - ord('a')] = 1
    length = max(length, r - l + 1)
    r += 1

print(length)
```

```python
s = "aabcbcdbca"
n = len(s)

distinct = len(set(s))

count = {}
l = 0
formed = 0
ans = n

for r in range(n):
    count[s[r]] = count.get(s[r], 0) + 1

    if count[s[r]] == 1:
        formed += 1

    while formed == distinct:
        ans = min(ans, r - l + 1)

        count[s[l]] -= 1
        if count[s[l]] == 0:
            formed -= 1
        l += 1

print(ans)
```

```python
s = 'AABCDAABCD'
n = len(s)

lps = [0] * n
pref = 0
suf = 1

while suf < n:
    if s[pref] == s[suf]:
        lps[suf] = pref + 1
        pref += 1
        suf += 1
    else:
        if pref == 0:
            lps[suf] = 0
            suf += 1
        else:
            pref = lps[pref - 1]

print(lps)
print(lps[n - 1])
```

```python
# reverse a string using recursion

s = "rohit"
s = list(s)   # convert to list
n = len(s)

def rev(s, start, end):
    if start >= end:
        return
    s[start], s[end] = s[end], s[start]
    rev(s, start + 1, end - 1)

rev(s, 0, n - 1)

s = ''.join(s)   # convert back to string
print(s)
```

```python
# lower to upper using recursion

s = "riya"
s = list(s)         
n = len(s)

def lowertoupper(s, index):
    if index == -1:
        return
    
    s[index] = chr(ord(s[index]) - 32)
    lowertoupper(s, index - 1)

lowertoupper(s, n - 1)

s = ''.join(s)     
print(s)
```
