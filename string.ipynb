```python
#reverse a string
s='aabbbaabbcccs'
s=list(s)
n=len(s)
start=0
end=n-1
while(start<end):
    s[start],s[end]=s[end],s[start]
    start+=1
    end-=1
s=''.join(s)
print(s)
```

```python
#palindrome

s='madam'
s=list(s)
n=len(s)
flag=True
start=0
end=n-1
while(start<end):
    if s[start]!=s[end]:
        flag=False
        print("not a palindrome")
        break
    
    start+=1
    end-=1

if flag:
    print("palindrome")
  
```

```python
s='1.1.1.1'
i=0
n=len(s)
ans=''
while(i<n):
    if s[i]=='.':
        ans+='[.]'
    else:
        ans+=s[i]
    i+=1

print(ans)
```

```python
#rotate string in anticlockwise direction
s='amazon'
n=len(s)
s=list(s)
temp=s[0]
for i in range(1,n):
    s[i-1]=s[i]

s[n-1]=temp
s=''.join(s)
print(s)

```

```python
#check if string is pangram
s="thequickbrownfoxjumpsoverthelazydog"
n=len(s)
ans=[0]*26
for i in range(n):
    indx=ord(s[i])-ord('a')
    ans[indx]=1

flag=True
for i in range(26):
    if(ans[i]==0):
        flag=False
        print("not a pangram")
        break

if flag:
    print("pangram")
```

```python
#roman to integer
s="XXIV"
# s=list(s)
sum=0
index=0
n=len(s)

def num(c):
    if c=='I':
        return 1
    elif c=='V':
        return 5
    elif c=='X':
        return 10
    elif c=='L':
        return 50
    elif c=='C':
        return 100
    elif c=='D':
        return 500
    else:
        return 1000



while(index<n-1):
    if (num(s[index])<num(s[index+1])):
        sum-=num(s[index])
    else:
        sum+=num(s[index])
    index+=1
sum+=num(s[n-1])

print(sum)
```
