```python
#longest sbaray with k
#Method1---------

arr=[1,2,3,1,1,1,2,4,2,3]
n=len(arr)
l=0
k=4
for i in range(n):
    sum=0
    for j in range(i,n):
        sum+=arr[0]
        if sum==k:
            l=max(l,j-i+1)

print(l)
```

```python
#Method --------2
arr=[1,2,3,1,1,1,2,4,2,3]
n=len(arr)
k=4
left,right=0,0
sum=arr[0]
maxl=0
while(right<n):
    while(left<=right and sum>k):
        sum-=arr[left]
        left+=1

    if sum==k:
        maxl=max(maxl,right-left+1)
        
    right+=1

    if(right<n):
        sum+=arr[right]
    
print(maxl)

```

```python
#Two sum problem 
#Methd 1
arr=[2,6,5,8,11]
target=14
n=len(arr)
for i in range(n):
    for j in range(i+1,n):
        if arr[i]+arr[j]==target:
            print("firs value",arr[i]," second value", arr[j])

```

```python
#Method 2

arr=[2,6,5,8,11]
target=14
n=len(arr)
left=0
right=n-1

while(left<right):
    sum=arr[left]+arr[right]
    if sum==target:
        print(left, right)
        break

    elif(sum<target):
        left+=1

    else:
        right-=1
        
```

```python
#Sort an array of 0,1,2

#Method 1

arr=[0,1,2,0,1,2,1,2,0,0,0,1]
n=len(arr)
count0=0
count1=0
count2=0
for i in range(n):
    if arr[i]==0:count0+=1
    if arr[i]==1:count1+=1
    else:count2+=1

for i in range(0,count0):arr[i]=0
for i in range(count0,count0+count1):arr[i]=1
for i in range(count0+count1,n):arr[i]=2

print(arr)
```

```python
#Method 2

#Dutch National Flag Algoritm

arr=[0,1,2,0,1,2,1,2,0,0,0,1]
n=len(arr)
low=0
mid=0
high=n-1
while(mid<=high):
    if arr[mid]==0:
        arr[low],arr[mid]=arr[mid],arr[low]
        low+=1
        mid+=1
    elif arr[mid]==1:
        mid+=1
    elif arr[mid]==2:
        arr[mid],arr[high]=arr[high],arr[mid]
        high-=1

print(arr)
```

```python
#majority element-------method 1

arr=[2,3,3,3,2,2,2,2,2,2,2,2,3,2,3,3,1,2,2]
n=len(arr)
for i in range(n):
    count=0
    for j in range(n):
        if arr[i]==arr[j]:
            count+=1
    if count>(n/2):
        print(arr[i])
        break
```

```python
#majrity element -----------method 2--Moores Voting algorithm

arr=[2,1,3,3,3,2,2,2,3,2,1,2,2]
n=len(arr)
count=0
for i  in range(n):
    if count==0:
        count=1
        ele=arr[i]
    elif arr[i]==ele:
        count+=1
    else:
        count-=1


length=0
for i in range(n):
    if arr[i]==ele:
        length+=1
if length>(n/2):
    print(ele,"of ",length)
else:
    print("no majority element")
```

```python
#maximum subarray------method 1
arr=[-2,-3,4,-1,-2,1,5,-3]
n=len(arr)
maxi=float("-inf")
for i in range(n):
    sum_val=0
    for j in range(i,n):
        sum_val=sum_val+arr[j]
        maxi=max(sum_val,maxi)

print("maximum subarray is ",maxi)
```

```python
#maximum subarray------method 2---kadanes algorithm
arr=[-2,-3,4,-1,-2,1,5,-3]
n=len(arr)
maxi=float("-inf")
sum=0
ansstart=-1
ansend=-1
for i in range(n):
    if sum==0:
        start=i
    sum=sum+arr[i]
    if sum>maxi:
        maxi=sum
        ansstart=start
        ansend=i
    if sum<0:
        sum=0

print("maximum val", maxi)
```

```python
#Best time to buy and sell stock

arr=[7,12,5,3,16,4,1]
n=len(arr)
mini=arr[0]
profit=0
for i in range(n):
    cost=arr[i]-mini
    profit=max(profit,cost)
    mini=min(mini,arr[i])
print(profit)
```

```python
#Rearrange element by sign
arr=[3,1,-2,-5,2,-4]
n=len(arr)
pos=[]
neg=[]
for i in range(n):
    if arr[i]>0:
        pos.append(arr[i])
    else:
        neg.append(arr[i])
for i in range(n//2):
    arr[2*i]=pos[i]
    arr[2*i+1]=neg[i]
print(arr)
```

```python
#leaders in python--------method 1
arr=[10,22,12,3,0,6]
n=len(arr)
temp=[]
for i in range(n):
    leader=True
    for j in range(i+1,n):
        if arr[j]>arr[i]:
            leader=False
            break
    if leader==True:
        temp.append(arr[i])
print(temp)
```

```python
#leaders in python
arr=[10,22,12,3,0,6]
n=len(arr)
temp=[]
max_value=float("-inf")

for i in range(n-1,-1,-1):
    if arr[i]>max_value:
        temp.append(arr[i])
        max_value=arr[i]

print(temp)
```

```python
#majority element 2--------brute approach

arr=[1,1,1,4,4,4,2,2]
n=len(arr)
ls=[]
for i in range(n):
    if len(ls)==0 or ls[0]!=arr[i]:
        count=0
        for j in range(n):
            if arr[j]==arr[i]:
                count+=1
        if count>n//3:
            ls.append(arr[i])

    if len(ls)==2:
        break

print(ls)
```

```python
#majority element 2--------optimal approach

arr=[1,1,1,3,3,3,2,2]
n=len(arr)
count1=0
count2=0
ele1=None
ele2=None

for i in range(n):
    if count1==0 and arr[i]!=ele2:
        count1=1
        ele1=arr[i]
    
    elif count2==0 and arr[i]!=ele1:
        count2=1
        ele2=arr[i]

    elif arr[i]==ele1:
        count1+=1
    elif arr[i]==ele2:
        count2+=1

    else:
        count1-=1
        count2-=1

count1=0
count2=0
l=n//3
ls=[]

for i in range(n):
    if arr[i]==ele1:
        count1+=1
    if arr[i]==ele2:
        count2+=1

if count1>l:
    ls.append(ele1)
if count2>l:
    ls.append(ele2)

print(ls)

```
