```python
#iterative binary search

arr=[3,4,5,6,8,9,10]
n=len(arr)
target=6
flag=False
low=0
high=n-1
while(low<=high):
    mid=(low+high)//2
    if arr[mid]==target:
        flag=True
        print("at index",mid)
        break
    elif target>arr[mid]:
        low=mid+1
    else:
        high=mid-1
if not flag:
    print("not in array")
```
