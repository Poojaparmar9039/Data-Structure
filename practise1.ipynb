## sorting

```python
#bubble sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
for i in range(n):
    for j in range(n-i-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

```python
#insertion sort
arr=[2,76,3,22,9,4,21,0,30,1]
print(arr)
n=len(arr)
for i in range(n):
    j=i
    while(j>0 and arr[j-1]>arr[j]):
        arr[j],arr[j-1]=arr[j-1],arr[j]
        j-=1
print(arr)
```

```python
#selection sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
for i in range(n):
    mini=i
    for j in range(i,n):
        if arr[j]<arr[mini]:
            mini=j
    arr[i],arr[mini]=arr[mini],arr[i]
print(arr)
```

```python
#merge sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
def mergesort(arr,low,high):
    if low>=high:return
    mid=(low+high)//2
    mergesort(arr,low,mid)
    mergesort(arr,mid+1,high)
    merge(arr,low,mid,high)

def merge(arr,low,mid,high):
    left=low
    right=mid+1
    temp=[]
    while(left<=mid and right<=high):
        if(arr[left]<=arr[right]):
            temp.append(arr[left])
            left+=1
        else:
            temp.append(arr[right])
            right+=1

    while(left<=mid):
        temp.append(arr[left])
        left+=1
    while(right<=high):
        temp.append(arr[right])
        right+=1
    
    for i in range(len(temp)):
        arr[low+i]=temp[i]

mergesort(arr,0,n-1)
print(arr)
```

```python
#quick sort

arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
def quicksort(arr,low,high):
    if(low>=high):
        return
    partition=findpivot(arr,low,high)
    quicksort(arr,low,partition-1)
    quicksort(arr,partition+1,high)

def findpivot(arr,low,high):
    i=low
    j=high
    pivot=arr[low]
    while i<j:
        while(i<=high and arr[i]<=pivot):
            i+=1
        while(j>=low and arr[j]>pivot):
            j-=1

        if i<j:
            arr[i],arr[j]=arr[j],arr[i]
    arr[low],arr[j]=arr[j],arr[low]
    return j
quicksort(arr,0,n-1)
print(arr)
```

```python
#linear search
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
key=int(input("enter value to search"))
def linearsearch(arr,n,key):
    for i in range(n):
        if arr[i]==key:
            print("yes key found at",i)
            return 
    return -1

linearsearch(arr,n,key)
```

```python
#binary search
arr=[2,76,3,22,9,4,21,0,30,1]
arr.sort()
n=len(arr)
key=int(input("enter value to search"))
def binarysearch(arr,low,high,key):
    if(low>=high):
        print("key not found")
        return
    mid=(low+high)//2
    if arr[mid]==key:
        print("found at",mid)

    if key<arr[mid]:
        binarysearch(arr,low,mid,key)
    elif key>arr[mid]:
        binarysearch(arr,mid+1,high,key)
    
binarysearch(arr,0,n-1,key)
print(arr)
    
```
