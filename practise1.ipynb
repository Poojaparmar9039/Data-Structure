```python
#bubble sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
for i in range(n):
    for j in range(n-i-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

```python
#insertion sort
arr=[2,76,3,22,9,4,21,0,30,1]
print(arr)
n=len(arr)
for i in range(n):
    j=i
    while(j>0 and arr[j-1]>arr[j]):
        arr[j],arr[j-1]=arr[j-1],arr[j]
        j-=1
print(arr)
```

```python
#selection sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
for i in range(n):
    mini=i
    for j in range(i,n):
        if arr[j]<arr[mini]:
            mini=j
    arr[i],arr[mini]=arr[mini],arr[i]
print(arr)
```

```python
#merge sort
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
def mergesort(arr,low,high):
    if low>=high:return
    mid=(low+high)//2
    mergesort(arr,low,mid)
    mergesort(arr,mid+1,high)
    merge(arr,low,mid,high)

def merge(arr,low,mid,high):
    left=low
    right=mid+1
    temp=[]
    while(left<=mid and right<=high):
        if(arr[left]<=arr[right]):
            temp.append(arr[left])
            left+=1
        else:
            temp.append(arr[right])
            right+=1

    while(left<=mid):
        temp.append(arr[left])
        left+=1
    while(right<=high):
        temp.append(arr[right])
        right+=1
    
    for i in range(len(temp)):
        arr[low+i]=temp[i]

mergesort(arr,0,n-1)
print(arr)
```

```python
#quick sort

arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
def quicksort(arr,low,high):
    if(low>=high):
        return
    partition=findpivot(arr,low,high)
    quicksort(arr,low,partition-1)
    quicksort(arr,partition+1,high)

def findpivot(arr,low,high):
    i=low
    j=high
    pivot=arr[low]
    while i<j:
        while(i<=high and arr[i]<=pivot):
            i+=1
        while(j>=low and arr[j]>pivot):
            j-=1

        if i<j:
            arr[i],arr[j]=arr[j],arr[i]
    arr[low],arr[j]=arr[j],arr[low]
    return j
quicksort(arr,0,n-1)
print(arr)
```

```python
#linear search
arr=[2,76,3,22,9,4,21,0,30,1]
n=len(arr)
key=int(input("enter value to search"))
def linearsearch(arr,n,key):
    for i in range(n):
        if arr[i]==key:
            print("yes key found at",i)
            return 
    return -1

linearsearch(arr,n,key)
```

```python
#binary search
arr=[2,76,3,22,9,4,21,0,30,1]
arr.sort()
n=len(arr)
key=int(input("enter value to search"))
def binarysearch(arr,low,high,key):
    if(low>=high):
        print("key not found")
        return
    mid=(low+high)//2
    if arr[mid]==key:
        print("found at",mid)

    if key<arr[mid]:
        binarysearch(arr,low,mid,key)
    elif key>arr[mid]:
        binarysearch(arr,mid+1,high,key)
    
binarysearch(arr,0,n-1,key)
print(arr)
    
```

```python
#left rotate by d places
arr=[1,2,3,4,5,6,7]
n=len(arr)
temp=[]
d=int(input())
d= d % n
for i in range(d):
    temp.append(arr[i])

for i in range(d,n):
    arr[i-d]=arr[i]
for i in range(n-d,n):
    arr[i]=temp[i-(n-d)]
print(arr)
```

```python
#optimal approach
arr=[1,2,3,4,5,6,7,8]
n=len(arr)
d=int(input())
d=d%n
def reverse(arr,start,end):
    while start<=end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
reverse(arr,0,d-1)
reverse(arr,d,n-1)
reverse(arr,0,n-1)
print(arr)
```

```python
#union of two sorted array
arr1=[1,1,2,3,6]
arr2=[0,1,2,2,4,5,7,7]
i=0
j=0
n=len(arr1)
m=len(arr2)
union=[]
while i<n and j<m:
    if arr1[i]<arr2[j]:
        if not union or union[-1]!=arr1[i]:
            union.append(arr1[i])
        i+=1
    elif arr1[i]>arr2[j]:
        if not union or union[-1]!=arr2[j]:
            union.append(arr2[j])
        j+=1
    else:
        if not union or union[-1]!=arr1[i]:
            union.append(arr1[i])
        i+=1
        j+=1
while(i<n):
    if not union or union[-1]!=arr1[i]:
         union.append(arr1[i])
    i+=1
while(j<m):
    if not union or union[-1]!=arr2[j]:
         union.append(arr2[j])
    j+=1 
print(union)
```

```python
#sort an array of 0,1,2
arr=[0,1,2,0,1,2,1,2,0,0,0,1]
n=len(arr)
count0=0
count1=0
count2=0
for i in range(n):
    if arr[i]==0:
        count0+=1
    elif arr[i]==1:
        count1+=1
    else:
        count2+=1
for i in range(count0):arr[i]=0
for i in range(count0,count0+count1):arr[i]=1
for i in range(count0+count1,n):arr[i]=2
print(arr)
```

```python
#sort an array of 0,1,2
arr=[0,1,2,0,1,2,1,2,0,0,0,1]
n=len(arr)
low=0
mid=0
high=n-1
while(mid<=high):
    if arr[mid]==0:
        arr[low],arr[mid]=arr[mid],arr[low]
        low+=1
        mid+=1
    elif arr[mid]==1:
        mid+=1
    elif arr[mid]==2:
        arr[mid],arr[high]=arr[high],arr[mid]
        high-=1
print(arr)
```

```python
#majority element--Moores oting algorithm
arr=[1,2,1,1,2,4,4,5,4,4,4,4,4,4,3]
n=len(arr)
count=0
for i in range(n):
    if count==0:
        count=1
        ele=arr[i]
    elif arr[i]==ele:
        count+=1
    else:
        count-=1

l=0
for i in range(n):
    if arr[i]==ele:
        l+=1
if l>n//2:
    print(ele)
```

```python
#maximum subarray------method 1
arr=[-2,-3,4,-1,-2,1,5,-3]
n=len(arr)
maxi=float("-inf")
for i in range(n):
    sum=0
    for j in range(i,n):
        sum+=arr[j]
        maxi=max(sum,maxi)
print(maxi)
```

```python
#maximum subarray------kadane's algorithm
arr=[-2,-3,4,-1,-2,1,5,-3]
n=len(arr)
maxi=float("-inf")
start=0
ansstart=-1
ansend=-1
sum=0
for i in range(n):
    if sum==0:
        start=i
    sum=sum+arr[i]
    if sum>maxi:
        maxi=sum
        ansstart=start
        ansend=i
    elif sum<0:
        sum=0

print(maxi)
```

```python
#best time to buy and sellstock
arr=[7,1,5,3,6,14]
n=len(arr)
mini=arr[0]
profit=0
for i in range(n):
    cost=arr[i]-mini
    profit=max(profit,cost)
    mini=min(mini,arr[i])
print(profit)
```

```python
#Rearrange element by sign
arr=[3,1,-2,-5,2,-4]
n=len(arr)
pos=[]
neg=[]
for i in range(n):
    if arr[i]>0:
        pos.append(arr[i])
    else:
        neg.append(arr[i])

for i in range(n//2):
    arr[i*2]=pos[i]
    arr[2*i+1]=neg[i]

print(arr)
```
