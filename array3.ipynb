```python
#3 sum ----------brute aproach

arr=[-1,0,1,2,-1,-4]
ans=[]
n=len(arr)
for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):
            if(arr[i]+arr[j]+arr[k]==0):
                triplet = sorted([arr[i], arr[j], arr[k]])
                if triplet not in ans:
                    ans.append(triplet)
                
                
print(ans)
```

```python
#4 sum brute approach
arr=[1,0,-1,0,-2,2]
target=0
n=len(arr)
ans=[]
for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):
            for l in range(k+1,n):
                sum=arr[i]+arr[j]+arr[k]+arr[l]
                if sum==target:
                    fourlet=sorted([arr[i],arr[j],arr[k],arr[l]])
                    if fourlet not in ans:
                        ans.append(fourlet)
print(ans)
```

```python
arr=[4,2,2,6,4]
n=len(arr)
k=6
count=0
for i in range(n):
    xor=0
    for j in range(i,n):
        xor=xor^arr[j]
        if xor==k:
            count+=1
print(count)
```

```python
#constant window
arr = [-1, 2, 3, 3, 4, 5, -1]
n = len(arr)
k = 4


window_sum = 7
max_sum = window_sum

l = 0
r = k

while r < n:
    window_sum = window_sum - arr[l] + arr[r]
    l += 1
    r += 1
    max_sum = max(max_sum, window_sum)

print(max_sum)
```

```python
# Fruit in basket

arr=[3,3,3,1,2,1,1,2,3,3,4]
n=len(arr)
maxlen=0

for i in range(n):
    s=set()
    for j in range(i,n):
        s.add(arr[j])
        if(len(s)<=2):
            maxlen=max(maxlen,j-i+1)
        else:
            break
print(maxlen)
        
```

```python
from collections import defaultdict

arr = [3,3,3,1,2,1,1,2,3,3,4]
n = len(arr)

l = 0
maxlen = 0
k = 2
mp = defaultdict(int)

for r in range(n):
    mp[arr[r]] += 1

    while len(mp) > k:
        mp[arr[l]] -= 1
        if mp[arr[l]] == 0:
            del mp[arr[l]]
        l += 1

    maxlen = max(maxlen, r - l + 1)

print(maxlen)
```
