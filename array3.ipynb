```python
#3 sum ----------brute aproach

arr=[-1,0,1,2,-1,-4]
ans=[]
n=len(arr)
for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):
            if(arr[i]+arr[j]+arr[k]==0):
                triplet = sorted([arr[i], arr[j], arr[k]])
                if triplet not in ans:
                    ans.append(triplet)
                
                
print(ans)
```

```python
#4 sum brute approach
arr=[1,0,-1,0,-2,2]
target=0
n=len(arr)
ans=[]
for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):
            for l in range(k+1,n):
                sum=arr[i]+arr[j]+arr[k]+arr[l]
                if sum==target:
                    fourlet=sorted([arr[i],arr[j],arr[k],arr[l]])
                    if fourlet not in ans:
                        ans.append(fourlet)
print(ans)
```

```python
#constant window
arr = [-1, 2, 3, 3, 4, 5, -1]
n = len(arr)
k = 4


window_sum = 7
max_sum = window_sum

l = 0
r = k

while r < n:
    window_sum = window_sum - arr[l] + arr[r]
    l += 1
    r += 1
    max_sum = max(max_sum, window_sum)

print(max_sum)
```

```python
from collections import defaultdict

arr = [3,3,3,1,2,1,1,2,3,3,4]
n = len(arr)

l = 0
maxlen = 0
k = 2
mp = defaultdict(int)

for r in range(n):
    mp[arr[r]] += 1

    while len(mp) > k:
        mp[arr[l]] -= 1
        if mp[arr[l]] == 0:
            del mp[arr[l]]
        l += 1

    maxlen = max(maxlen, r - l + 1)

print(maxlen)
```

```python
s = "bbacba"
n = len(s)
count = 0

for i in range(n):
    freq = [0, 0, 0]   # for a, b, c
    for j in range(i, n):
        freq[ord(s[j]) - ord('a')] = 1
        if freq[0] + freq[1] + freq[2] == 3:
            count += n-j
            break
print(count)
```

```python
s = "bbacba"
n = len(s)
count = 0

freq = [-1, -1, -1]   # last seen index of a, b, c

for i in range(n):
    freq[ord(s[i]) - ord('a')] = i

    if freq[0] != -1 and freq[1] != -1 and freq[2] != -1:
        count += 1 + min(freq[0], freq[1], freq[2])

print(count)
```

```python
# longest repeating character replacement (brute force)

s = "AABABBA"
k = 1
n = len(s)
maxlen = 0

for i in range(n):
    hash = [0] * 26
    maxf = 0
    for j in range(i, n):
        idx = ord(s[j]) - ord('A')
        hash[idx] += 1
        maxf = max(maxf, hash[idx])

        changes = (j - i + 1) - maxf
        if changes <= k:
            maxlen = max(maxlen, j - i + 1)
        else:
            break

print(maxlen)
```

```python
arr = [1,0,1,0,1]
n = len(arr)
k = 2

def atMost(arr, k):
    l = 0
    sum = 0
    count = 0   
    for r in range(len(arr)):
        sum += arr[r]  
        while sum > k:
            sum -= arr[l]
            l += 1    
        count += (r - l + 1)
    return count

ans = atMost(arr, k) - atMost(arr, k-1)
print(ans)
```

```python
arr = [2,1,1,1,3,4,3,2]
k = 3

def func(arr, k):
    count = 0
    l = 0
    r = 0
    map = {}

    while r < len(arr):
        map[arr[r]] = map.get(arr[r], 0) + 1

        while len(map) > k:
            map[arr[l]] -= 1
            if map[arr[l]] == 0:
                del map[arr[l]]
            l += 1

        count += (r - l + 1)
        r += 1

    return count

ans = func(arr, k) - func(arr, k-1)
print(ans)
```
